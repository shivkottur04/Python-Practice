```python
#bubble sort
num=[10,5,3,6,7]
n=len(num)
for i in range(n):
    for j in range(0,n-1-i):
        if num[j]>num[j+1]:
            num[j],num[j+1]=num[j+1],num[j]
print(num)
```

```python
#selection sort
num=[10,5,3,7,6]
n=len(num)
for i in range(n-1):
    minimum=i
    for j in range(i+1,n):
        if num[j]<num[minimum]:
            minimum=j
    num[i],num[minimum]=num[minimum],num[i]
print(num)
```

```python
#insertion sort
num=[10,5,3,7,6]
n=len(num)
for i in range(n):
    key=num[i]
    j=i-1
    while j>=0 and key<num[j]:
        num[j+1]=num[j]
        j-=1
    num[j+1]=key
print(num)
```
