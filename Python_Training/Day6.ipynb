```python
#DOUBLY LINKED LIST
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        self.prev=None
class LinkedList:
    def __init__(self):
        self.head=None
    def insertAtBeg(self,data):
        newNode=Node(data)
        if self.head==None:
            self.head=newNode
        else:
            newNode.next=self.head
            self.head.prev=newNode
            self.head=newNode
    def insertAtEnd(self,data):
        newNode=Node(data)
        if self.head==None:
            self.head=newNode
        else:
            temp=self.head
            while temp.next:
                temp=temp.next
            temp.next=newNode
            newNode.prev=temp
    def floyds_detection_cycle(self):
        slow=self.head
        fast=self.head
        while fast and fast.next:
            slow=slow.next
            fast=fast.next.next
            if slow==fast:
                return True
        return False
    def display(self):
        temp=self.head
        while temp:
            print(temp.data,end="")
            print("->",end="")
            temp=temp.next
        print(None)
        
#find element through backwards
    def find_from_back(self,data):
        temp=self.head
        while temp.next:
            temp=temp.next
        count=1
        while temp:
            if temp.data==data:
                print(f"found at position {count} from backwards")
                return
            else:
                count+=1
                temp=temp.prev
        print("not found")
    
        
            
            
l=LinkedList()
l.insertAtBeg(10)
l.insertAtEnd(20)
l.insertAtEnd(30)
print(l.floyds_detection_cycle())
l.display()
l.find_from_back(20)

        
```
