```python
#even and odd elements
l=[1,2,3,4,5,6]
even_count=0
odd_count=0
for i in l:
    if i%2==0:
        even_count+=1
    else:
        odd_count+=1
print("even:",even_count)
print("odd:",odd_count)
```

```python
#reverse an array
lst=[1,2,3,4,5,6]
rev=[]
for i in range(len(lst)-1,-1,-1):
    rev.append(lst[i])
    lst.remove(lst[i])
print(rev)
```
