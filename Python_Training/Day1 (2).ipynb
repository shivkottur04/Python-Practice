```python
#fibonacci series without recurssion
n=int(input())   #n is no. of terms i.e 0 to (n-1)th term
lst=[0,1]
if n==0:
    print([0])
elif n==1:
    print(lst)
else:
    for i in range(2,n):
        lst.append(lst[i-1]+lst[i-2])
    print(lst)
        
```

```python
#### 7.
bank={"1234567890":"1234","9087654321":"5678"}
balance=1000      #default
print("----------Login!!-----------")
acc=input("enter account number:")
pin=input("enter your pin")
if acc in bank.keys():
    if pin==bank[acc]:
        print("login successful")
        while True:
            print("1.deposit\n2.withdraw\n3.check balance\n4.exit")
            ch=int(input("enter your choice:"))
            if ch==1:
                amt=int(input("Enter amount to deposit:"))
                balance+=amt
                print("deposit successful")
            elif ch==2:
                amt=int(input("enter amount for withdrawal:"))
                if amt<=balance:
                    balance-=amt
                    print("withdrawal successful:")
                else:
                    print("not enough balance \n Try again")
            elif ch==3:
                print("account balance:",balance)
            elif ch==4:
                print("thank you!")
                break
            else:
                print("invalid choice")
        else:
             print("wrong pin!!Try again later")
else:
       print("invalid account number")



```

```python
#9:
def check_upper(x):
    if x.isupper():
        return 1
    return 0
def check_special(x):
    if not(x.isupper()) and not(x.islower()) and not(x.isdigit()):
        return 1
    return 0
def check_digit(x):
    if x.isdigit():
        return 1
    return 0
def check_lower(x):
    if x.islower():
        return 1
    return 0
u=input("create username")


flag=0
while flag==0:
    p=input("enter password")
    up_count=0
    low_count=0
    spl_count=0
    
    digit_count=0
    
    for i in p:
        if check_upper(i):
            up_count+=1
        elif check_lower(i):
            low_count+=1
        elif check_digit(i):        
            digit_count+=1
        elif check_special(i):
            spl_count+=1
    if up_count>=1 and low_count>=1 and digit_count>=1 and spl_count>=1 and len(p)>8:
        flag=1
    else:
        print("invalid!! Try again")
print("password valid")
        
            

```
