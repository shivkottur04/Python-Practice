```python
#single inheritance
class person:
    def __init__(self,name,age,blood_group,ph_no):
        self.name=name
        self.age=age
        self.blood_group=blood_group
        self.ph_no=ph_no
class employee(person):
    def __init__(self,name,age,blood_group,ph_no,empid,salary,department):
        super().__init__(name,age,blood_group,ph_no)
        self.empid=empid
        self.salary=salary
        self.department=department
    def display(self):
        print("name=",self.name)
        print("age=",self.age)
        print("blood group=",self.blood_group)
        print("phone number=",self.ph_no)
        print("employee id=",self.empid)
        print("salary",self.salary)
        print("department=",self.department)
s=employee("shiv",20,"b+",8217477249,101,900000,"finance")
s.display()
        
```

```python
#multi level inheritance
class vehicle:
    def __init__(self,wheeler):
        self.wheeler=wheeler
    def display(self):
        print("wheeler:",self.wheeler)
class four_wheeler(vehicle):
    def __init__(self,wheeler,Type):
        super().__init__(wheeler)
        self.Type=Type
    def display(self):
        super().display()
        print("type of vehicle:",self.Type)
class car(four_wheeler):
    def __init__(self,wheeler,Type,company,model,year):
        super().__init__(wheeler,Type)
        self.company=company
        self.model=model
        self.year=year
    def display(self):
        super().display()
        print("company:",self.company)
        print("model:",self.model)
        print("year:",self.year)
c=car("four wheeler","car","ford","classic",2013)
c.display()
        
    
```

```python
#multiple inheritance
class father:
    def fdisplay(self):
        print("i am father")
class mother:
    def mdisplay(self):
        print("i am mother")
class child(father,mother):
    def display(self):
        print("i am the child")
c=child()
c.fdisplay()
c.mdisplay()
c.display()
```

```python
#wap to create class employee with attribute name and salary create another class developer that inherits property of class employee 
#and again create another class of HR that will inherit property of class employee and developer.display all the details(multi level inheritance)
class employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
class developer(employee):
    def __init__(self,name,salary,skills):
        super().__init__(name,salary)
        self.skills=skills
class HR(developer):
    def __init__(self,name,salary,skills,manages):
        super().__init__(name,salary,skills)
        self.manages=manages
    def display(self):
        print("name:",self.name)
        print("salary:",self.salary)
        print("skill:",self.skills)
        print("manages:",self.manages)
h=HR("shiv",1000000,"python","projects")
h.display()
```

```python
#hybrid inheritance
class vehicle:
    def __init__(self,wheeler):
        self.wheeler=wheeler
    def display(self):
        print("wheeler:",self.wheeler)
class four_wheeler(vehicle):
    def __init__(self,wheeler,Type):
        super().__init__(wheeler)
        self.Type=Type
    def display(self):
        super().display()
        print("type of vehicle:",self.Type)
class car(four_wheeler):
    def __init__(self,wheeler,Type,company,model,year):
        super().__init__(wheeler,Type)
        self.company=company
        self.model=model
        self.year=year
    def display(self):
        super().display()
        print("company:",self.company)
        print("model:",self.model)
        print("year:",self.year)
class truck(four_wheeler):
    def __init__(self,wheeler,Type,company,model,year):
        super().__init__(wheeler,Type)
        self.company=company
        self.model=model
        self.year=year
    def display(self):
        super().display()
        print("company:",self.company)
        print("model:",self.model)
        print("year:",self.year)
print("-----------car-------------")
c=car("four wheeler","car","ford","classic",2013)
c.display()
print("---------- truck-----------")
t=truck("four wheeler","truck","ashok leyland","blazo",2000)
t.display()

```

```python
#function overriding
class parent:
    def display(self):
        print("guna")
class child(parent):
    def display(self):
        #parent.display(self)
        #super().display()
        print("shiv")
c=child()
c.display()
```

```python
#static method
class parent:
    @staticmethod
    def display():
        print("guna")
class child(parent):
    @staticmethod
    def display():
        parent.display()
        print("shiv")
c=child()
c.display()
```

```python
#polymorphism
class Bird:
    def fly(self):
        print("birds can fly")
class Penguin(Bird):
    def fly(self):
        print("penguins cannot fly")
def flytest(bird):
    bird.fly()
sparrow=Bird()
penguin=Penguin()
flytest(sparrow)
flytest(penguin)
```

```python
#private variable
class BankAccount:
    def __init__(self,balance):
        self.__balance=balance  #private variable
    def deposit(self,amount):
        self.__balance+=amount
    def get_balance(self):
        return self.__balance
b=BankAccount(10000)
b.deposit(2000)
print(b.get_balance())
```
