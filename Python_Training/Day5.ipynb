```python
#find all pairs whose sum is equal to given number
lst=list(map(int,input().split()))
k=int(input("enter the target number:"))
pairs=[]
for i in range(len(lst)-1):
    for j in range(i+1,len(lst)):
        if lst[i]+lst[j]==k:
            pairs.append((lst[i],lst[j]))
print(pairs)

```

```python
#two pointer
lst=[10,20,30,40,50]
left=0
right=len(lst)-1
while left<right:
    lst[left],lst[right]=lst[right],lst[left]
    left+=1
    right-=1
print(lst)
```

```python
#palindrome using two pointer
number=input()
num=list(number)
left=0
right=len(num)-1
while left<right:
    num[left],num[right]=num[right],num[left]
    left+=1
    right-=1
rev="".join(num)
print(rev)
if number==rev:
    print("palindrome")
else:
    print("not palindrome")
```

```python
#remove duplicate from sorted array using two pointer

lst=[1,2,4,3,2,4,5,6]
lst.sort()
p1=0
p2=1
while p2<=len(lst)-1:
    if lst[p1]==lst[p2]:
        lst.remove(lst[p2])
    else:
        p1+=1
        p2+=1
print(lst)
```

```python
height = [4,2,0,3,2,5]

left = 0
right = len(height) - 1

left_max = 0
right_max = 0

water = 0

while left < right:

    if height[left] < height[right]:

        if height[left] >= left_max:
            left_max = height[left]
        else:
            water += left_max - height[left]

        left += 1

    else:

        if height[right] >= right_max:
            right_max = height[right]
        else:
            water += right_max - height[right]

        right -= 1

print(water)
```
